```python
from ultralytics import YOLO
from pathlib import Path
import matplotlib.pyplot as plt
import cv2
```

```python
# caminho do modelo treinado
model_path = r"modelo.pt" 
```

```python
# pasta com as novas imagens
source_path = r"novas_imagens"
```

```python
# carregar o modelo
model = YOLO(model_path)
```

```python
# rodar predição nas novas imagens
results = model.predict(
    source=source_path,
    conf=0.25,      # confiança mínima
    save=True,      # salva as imagens com as detecções
    imgsz=640       # tamanho da imagem
)

print(f"Predições realizadas em {len(results)} imagens.")
```

```python
img_path = r"novas_imagens/laranja3.png"

results = model.predict(
    source=img_path,
    conf=0.25,
    save=False,
    verbose=False
)
print(results)
# pega o primeiro resultado
img_pred = results[0].plot()

# corrigir cor
img_pred = cv2.cvtColor(img_pred, cv2.COLOR_BGR2RGB)

plt.imshow(img_pred)
```
